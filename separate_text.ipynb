```python
import pyperclip


def format_spasi(teks):
    kata = teks.split()
    # 1 spasi antar huruf,      
    # 3 spasi antar kata
    return "   ".join(" ".join(k) for k in kata)


teks = input("Masukkan kalimat: ")

hasil = format_spasi(teks)

print("\nHasil:")
print(hasil)

# Salin otomatis ke clipboard Windows
pyperclip.copy(hasil)

print("\n✓ Hasil berhasil disalin ke clipboard!")

# Verifikasi isi clipboard
print("\nIsi clipboard:")
print(pyperclip.paste())
```
